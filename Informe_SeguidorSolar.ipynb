# Resumen

El presente proyecto desarrolla un sistema computacional capaz de simular el comportamiento de un seguidor solar de dos grados de libertad. El objetivo principal consiste en calcular los ángulos de control necesarios para orientar un panel solar de forma perpendicular a la radiación solar incidente, maximizando así el aprovechamiento energético.

Para lograrlo, se utiliza la biblioteca Pysolar para obtener la posición del Sol mediante los ángulos de azimut y elevación. A partir de estos parámetros se construye un modelo matemático basado en vectores tridimensionales y matrices de rotación que permite determinar los ángulos mecánicos de control denominados roll y pitch.

El sistema implementado genera una simulación tridimensional interactiva que muestra simultáneamente la trayectoria solar y la orientación del panel durante un intervalo de tiempo seleccionado por el usuario.

## Objetivo Principal: 

Desarrollar una aplicación computacional que simule el funcionamiento de un seguidor solar de dos grados de libertad, capaz de calcular los ángulos de control necesarios para orientar un panel solar perpendicularmente a la radiación solar incidente, utilizando la posición solar obtenida a partir del azimut y la elevación para una fecha y ubicación geográfica determinadas.

### Objetivos Específicos:

* 1. Obtener la posición del Sol mediante los ángulos de azimut y elevación utilizando la biblioteca Pysolar.

* 2. Desarrollar el procedimiento matemático para calcular los ángulos de control roll y pitch a partir de la posición solar.

* 3. Verificar matemáticamente que la normal del panel coincida con la dirección de la radiación solar mediante el cálculo del error angular.


# 1. Introducción

La energía solar constituye una de las fuentes renovables más importantes debido a su disponibilidad, sostenibilidad y bajo impacto ambiental.

Los paneles solares instalados de forma fija presentan pérdidas energéticas debido a la variación continua de la posición solar a lo largo del día. Para solucionar este problema se emplean seguidores solares, dispositivos capaces de modificar automáticamente la orientación del panel para mantener una incidencia óptima de la radiación.

El presente proyecto desarrolla una simulación computacional de un seguidor solar de dos grados de libertad capaz de calcular la posición del Sol y determinar los ángulos de control necesarios para orientar el panel de forma perpendicular a la radiación solar incidente.


# 2. Marco Teórico

## 2.1 Posición Solar

### Azimut (α)

Representa el ángulo horizontal medido desde el norte geográfico hacia la proyección del Sol sobre el plano horizontal.

- 0° = Norte
- 90° = Este
- 180° = Sur
- 270° = Oeste

### Elevación (θ)

Representa el ángulo vertical del Sol respecto al horizonte.

- 0° = Horizonte
- 90° = Cenit

## 2.2 Vector Solar

La dirección de la radiación solar se representa mediante:

$$
\vec{s}=
\begin{bmatrix}
\cos(\theta)\sin(\alpha)\\
\cos(\theta)\cos(\alpha)\\
\sin(\theta)
\end{bmatrix}
$$

## 2.3 Matrices de Rotación

### Pitch

$$
R_x(p)=
\begin{bmatrix}
1&0&0\\
0&\cos p&-\sin p\\
0&\sin p&\cos p
\end{bmatrix}
$$

### Roll

$$
R_y(r)=
\begin{bmatrix}
\cos r&0&\sin r\\
0&1&0\\
-\sin r&0&\cos r
\end{bmatrix}
$$

## 2.4 Seguidores Solares de Dos Grados de Libertad

Permiten controlar simultáneamente dos movimientos independientes:
- Pitch
- Roll

Con ello el panel puede orientarse hacia cualquier posición visible del Sol.


# 3. Metodología

## Algoritmo 1: Obtención de la Posición Solar

1. Ingresar fecha y hora.
2. Obtener latitud y longitud.
3. Calcular azimut y elevación mediante Pysolar.
4. Retornar los valores calculados.

## Algoritmo 2: Construcción del Vector Solar

$$
\vec{s}=
\begin{bmatrix}
\cos(\theta)\sin(\alpha)\\
\cos(\theta)\cos(\alpha)\\
\sin(\theta)
\end{bmatrix}
$$

## Algoritmo 3: Cálculo de Roll y Pitch

Sea:

$$
\vec{s}=(x,y,z)
$$

Entonces:

$$
pitch=-\arcsin(y)
$$

$$
roll=\operatorname{atan2}(x,z)
$$

## Algoritmo 4: Reconstrucción de la Normal

$$
\vec{n}_0=
\begin{bmatrix}
0\\0\\1
\end{bmatrix}
$$

$$
\vec{n}=R_y(roll)R_x(pitch)\vec{n}_0
$$

## Algoritmo 5: Verificación

$$
Error=\cos^{-1}(\vec{s}\cdot\vec{n})
$$

Si el panel está correctamente orientado:

$$
Error \approx 0^\circ
$$


# 4. Análisis de Resultados

## 4.1 Trayectoria Solar

La simulación muestra el desplazamiento aparente del Sol a lo largo del día para la ubicación del Campus de la Escuela Politécnica Nacional.

Se observa que el azimut aumenta progresivamente desde el amanecer hasta el atardecer, mientras que la elevación alcanza un valor máximo cercano al mediodía solar.

## 4.2 Variación de Pitch y Roll

Los ángulos de control presentan un comportamiento continuo y suave.

Durante las primeras horas del día los valores de pitch y roll son elevados debido a la baja altura del Sol.

Conforme aumenta la elevación solar, ambos ángulos se ajustan automáticamente para mantener la perpendicularidad.

## 4.3 Error Angular

Los resultados muestran errores cercanos a:

0.000000°

lo que demuestra que la normal del panel coincide con la dirección de la radiación solar.

- Orientación del panel.

- Dirección de la radiación solar.
